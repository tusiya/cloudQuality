```python

import matplotlib.pyplot as plt
import numpy as np

criteria = [
    "Quality model",
    "Functionality",
    "Performance",
    "Satisfaction",
    "Reliability",
    "Flexibility",
    "Protection"
]

methods = [
    "System metrics + equal weights",
    "User surveys + AHP weights",
    "System metrics + AHP weights"
]

baseline = {
    methods[0]: [4.268, 4.820, 4.450, 3.211, 4.000, 4.126, 5.000],
    methods[1]: [3.523, 3.642, 3.626, 3.426, 3.500, 3.861, 3.529],
    methods[2]: [3.808, 4.792, 4.455, 3.212, 4.000, 4.199, 5.000],
}

# AGE COMPARISON
age = {
    "Everyone (n=34)": baseline,

    "18–24 (n=13)": {
        methods[0]: [4.268, 4.820, 4.450, 3.211, 4.000, 4.126, 5.000],
        methods[1]: [3.297, 3.260, 3.340, 3.292, 3.077, 3.625, 3.154],
        methods[2]: [3.793, 4.796, 4.448, 3.218, 4.000, 4.146, 5.000],
    },

    "25–34 (n=6)": {
        methods[0]: [4.268, 4.820, 4.450, 3.211, 4.000, 4.126, 5.000],
        methods[1]: [3.581, 3.736, 3.716, 3.377, 3.500, 4.339, 3.667],
        methods[2]: [3.832, 4.776, 4.465, 3.210, 4.000, 4.151, 5.000],
    },

    "35–44 (n=5)": {
        methods[0]: [4.268, 4.820, 4.450, 3.211, 4.000, 4.126, 5.000],
        methods[1]: [3.336, 3.847, 3.610, 3.157, 3.200, 3.584, 3.400],
        methods[2]: [3.814, 4.801, 4.448, 3.201, 4.000, 4.476, 5.000],
    },

    "45–54 (n=10)": {
        methods[0]: [4.268, 4.820, 4.450, 3.211, 4.000, 4.126, 5.000],
        methods[1]: [3.890, 4.008, 3.984, 3.775, 4.200, 3.972, 4.000],
        methods[2]: [3.807, 4.791, 4.462, 3.210, 4.000, 4.154, 5.000],
    },
}

# PROFESSIONAL GROUP COMPARISON
group = {
    "Everyone (n=34)": baseline,

    "Client meetings / remote work (n=21)": {
        methods[0]: baseline[methods[0]],
        methods[1]: [3.621, 3.779, 3.765, 3.510, 3.762, 3.896, 3.571],
        methods[2]: [3.817, 4.787, 4.454, 3.208, 4.000, 4.225, 5.000],
    },

    "Student (n=5)": {
        methods[0]: baseline[methods[0]],
        methods[1]: [2.863, 2.867, 3.000, 2.800, 2.800, 3.208, 2.800],
        methods[2]: [3.782, 4.785, 4.453, 3.230, 4.000, 4.053, 5.000],
    },
}

# USAGE FREQUENCY COMPARISON
frequency = {
    "Everyone (n=34)": baseline,

    "Daily or almost daily (n=22)": {
        methods[0]: baseline[methods[0]],
        methods[1]: [3.637, 3.788, 3.797, 3.531, 3.773, 3.943, 3.545],
        methods[2]: [3.814, 4.785, 4.454, 3.209, 4.000, 4.220, 5.000],
    },
}

# DIGITAL SKILLS COMPARISON
digital = {
    "Everyone (n=34)": baseline,

    "Bardzo wysoki (n=15)": {
        methods[0]: baseline[methods[0]],
        methods[1]: [3.438, 3.537, 3.595, 3.284, 3.667, 3.814, 3.467],
        methods[2]: [3.832, 4.789, 4.446, 3.210, 4.000, 4.198, 5.000],
    },

    "Wysoki (n=11)": {
        methods[0]: baseline[methods[0]],
        methods[1]: [3.681, 3.813, 3.663, 3.638, 3.636, 4.028, 3.455],
        methods[2]: [3.781, 4.781, 4.470, 3.205, 4.000, 4.178, 5.000],
    },

    "Średni (n=7)": {
        methods[0]: baseline[methods[0]],
        methods[1]: [3.821, 3.943, 3.959, 3.755, 3.286, 4.139, 4.143],
        methods[2]: [3.820, 4.799, 4.454, 3.221, 4.000, 4.293, 5.000],
    },
}


# PLOTTING FUNCTION
def plot_category(category_name, datasets):

    fig, axes = plt.subplots(
        1, 2,
        figsize=(17, 6),
        sharey=True
    )

    x = np.arange(len(criteria))

    baseline_method = methods[0]
    comparison_methods = methods[1:]

    for ax, method in zip(axes, comparison_methods):

        for i, (group_name, group_data) in enumerate(datasets.items()):

            # Black dotted baseline
            ax.plot(
                x,
                group_data[baseline_method],
                color="black",
                linestyle=":",
                linewidth=2,
                alpha=0.65,
                zorder=1,
                label=baseline_method if i == 0 else "_nolegend_"
            )

            # AHP comparison lines
            ax.plot(
                x,
                group_data[method],
                marker="o",
                linewidth=2,
                markersize=5,
                label=group_name,
                zorder=2
            )

        ax.set_title(method, fontsize=11)

        ax.set_xticks(x)
        ax.set_xticklabels(
            criteria,
            rotation=35,
            ha="right"
        )

        ax.set_ylim(1, 5.2)
        ax.set_ylabel("Score")
        ax.grid(True, alpha=0.3)
        ax.legend(fontsize=8)

    fig.suptitle(category_name, fontsize=16)
    fig.tight_layout()
    plt.show()


# DISPLAY ALL COMPARISONS
plot_category("Age comparison", age)

plot_category("Professional group comparison", group)

plot_category("Usage frequency comparison", frequency)

plot_category("Digital skills comparison", digital)
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path

# --------------------------------------------------
# Configuration
# --------------------------------------------------

DATA_DIR = Path("../../stats")
OUTPUT_DIR = Path("weight_plots")
OUTPUT_DIR.mkdir(exist_ok=True)

categories = {
    "Age": {
        1: "18–24",
        3: "25–34",
        4: "35–44",
        2: "45–54",
    },
    "User Group": {
        5: "Client meetings / remote work",
        6: "Student",
    },
    "Usage Frequency": {
        7: "Daily or almost daily",
    },
    "Digital Competence": {
        8: "Very high",
        9: "High",
        10: "Medium",
    },
}

# --------------------------------------------------
# Load data
# --------------------------------------------------

data = {}

for i in range(11):
    data[i] = pd.read_csv(
        DATA_DIR / f"weights{i}.csv",
        encoding="utf-8-sig"
    )

# --------------------------------------------------
# Translate hierarchy labels
# --------------------------------------------------

translations = {
    "Model jakości": "Quality Model",
    "Funkcjonalność": "Functionality",
    "Wydajność": "Performance Efficiency",
    "Satysfakcja użytkowników": "User Satisfaction",
    "Wiarygodność": "Reliability",
    "Elastyczność": "Flexibility",
    "Ochrona": "Security",

    "Kompletność funkcjonalna": "Functional Completeness",
    "Poprawność funkcjonalna": "Functional Correctness",
    "Wydajność czasowa": "Time Behaviour",
    "Efektywność interakcji": "Interaction Efficiency",
    "Łatwość użycia": "Ease of Use",
    "Zrozumiałość obsługi": "Interaction Clarity",
    "Łatwość opanowania obsługi": "Learnability",
    "Odporność na błędy użytkownika": "User Error Protection",
    "Niezawodność": "Reliability",
    "Integracyjność": "Interoperability",
    "Konfigurowalność": "Adaptability",
    "Poufność": "Confidentiality",

    "Pokrycie funkcjonalne": "Functional Coverage",
    "Prawidłowy czas reakcji": "Response Time",
    "Liczba kroków potrzebnych do wykonania zadania": "Number of Steps Required to Complete a Task",
    "Sukces zadania użytkownika": "User Task Success",
    "Możliwość monitorowania": "Monitorability",
    "Samoopis interfejsu użytkownika": "Self-Descriptiveness",
    "Ustawienie wartości domyślnych": "Default Values",
    "Możliwość anulowania działań": "Action Reversibility",
    "Ochrona przed błędami użytkownika": "User Error Prevention",
    "Dostępność usługi": "Service Availability",
    "Zgodność formatów danych": "Data Format Compatibility",
    "Funkcjonalna możliwość personalizacji": "Functional Customization",
    "Konfigurowalność interfejsu użytkownika": "User Interface Customizability",
    "Kontrola dostępu": "Access Control",
}

for i in data:
    data[i]["parent_name"] = data[i]["parent_name"].replace(translations)
    data[i]["child_name"] = data[i]["child_name"].replace(translations)

baseline = data[0].copy()

# --------------------------------------------------
# Plotting function
# --------------------------------------------------

def plot_category_group(group_name, group_files):

    files = {0: "All users"}
    files.update(group_files)

    # Keep only parents with at least two children
    parents = (
        baseline.groupby("parent_name")["child_name"]
        .nunique()
    )

    parents = parents[parents > 1].index.tolist()

    fig, axes = plt.subplots(
        len(parents),
        1,
        figsize=(16, 5 * len(parents)),
        squeeze=False
    )

    axes = axes.flatten()

    for ax, parent in zip(axes, parents):

        children = baseline.loc[
            baseline["parent_name"] == parent,
            "child_name"
        ].tolist()

        x = np.arange(len(children))
        width = 0.8 / len(files)

        for j, (file_id, label) in enumerate(files.items()):

            df = data[file_id]

            parent_data = df[
                df["parent_name"] == parent
            ]

            weights = (
                parent_data
                .set_index("child_name")["weight"]
                .reindex(children)
            )

            positions = (
                x + (j - (len(files) - 1) / 2) * width
            )

            ax.bar(
                positions,
                weights,
                width,
                label=label
            )

        ax.set_title(
            parent,
            fontsize=12,
            fontweight="bold"
        )

        ax.set_ylabel("Weight")

        ax.set_xticks(x)
        ax.set_xticklabels(
            children,
            rotation=35,
            ha="right"
        )

        # Independent automatic Y-axis scaling
        max_weight = ax.get_ylim()[1]
        ax.set_ylim(0, max_weight * 1.15)

        ax.grid(axis="y", alpha=0.25)
        ax.set_axisbelow(True)

        ax.legend()

    fig.tight_layout()

    filename = group_name.lower().replace(" ", "_")

    fig.savefig(
        OUTPUT_DIR / f"{filename}.png",
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)


# --------------------------------------------------
# Generate all comparison figures
# --------------------------------------------------

for group_name, group_files in categories.items():
    plot_category_group(group_name, group_files)

print(f"All plots saved in: {OUTPUT_DIR.resolve()}")
```
